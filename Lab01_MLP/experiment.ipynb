```python
%load_ext autoreload
%autoreload 2

from src.mnsti_classifier import run_pipeline
import matplotlib.pyplot as plt
import numpy as np
```

```python
BATCH_SIZE = 32
LR = 0.01
OPTIM = "SGD"
SIZE_TRAIN = 50000
EPOCHS = 20
```

```python
train_loss, val_loss, train_accuracy, test_accuracy = run_pipeline(batch_size=BATCH_SIZE,
                                                         lr=LR,
                                                         optim_name=OPTIM,
                                                         epochs=EPOCHS,
                                                         size_train=SIZE_TRAIN)
```

```python
print(f"Epochs recorded: {len(train_loss)} (expected {EPOCHS})")
print(f"Train accuracy: {train_accuracy:.2f}%")
print(f"Test accuracy: {test_accuracy:.2f}%")
```

```python
plt.title("Loss curves")
plt.plot(train_loss)
plt.plot(val_loss)
plt.xticks(ticks=np.arange(1, stop=EPOCHS, step=5))
plt.xlabel("Epochs")
plt.ylabel("Mean loss")
plt.legend(["Train loss", "Validation loss"])
```
